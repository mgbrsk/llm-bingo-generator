# Локальная llm

```python
from src.models.llm_processor import UrlLlmProcessor
```

```python
llm = UrlLlmProcessor(model_id="deepseek-r1-distill-qwen-7b", temperature=0.7)
```

```python
llm.get_model_list()
```

```python
# Define the chat input
messages_list = [
    {
        "role": "system",
        "content": "Ты отвечаешь только на русском языке. Ты составитель вариантов для бинго. Варианты должны быть забавными. 5 вариантов, разделитель запятая.",
    },
    {
        "role": "user",
        "content": "Чем заняться весной? 4 варианта, через запятую",
    },
]

response = llm.get_completions_for_messages(messages_list)
print(response["answer"])
```

```python
# Define the chat input
messages_list = [
    {
        "role": "system",
        "content": "Ты форматируешь текст. Убери все цифры из текста, переведи всё на русский. Доступный ответ ТОЛЬКО объект type(list)",
    },
    {
        "role": "user",
        "content": "1. Пить молоко на улице  "
        "2. Смотреть луна вечером"
        "3. Пить чай и смотреть утренние сны  "
        "4. Наслаждаться солнечными лучами мамой  "
        "5. Пить и спать ",
    },
]
response = llm.get_completions_for_messages(messages_list)
print(response["answer"])
```
